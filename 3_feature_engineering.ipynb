# 3. Feature engineering
Objetivo: transformar variables categoricas y normalizar Age y Fare.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import QuantileTransformer, MinMaxScaler
```

## Carga de datos

```python
df = pd.read_csv('./data/titanic_clean.csv')
df.head()
```

## Encoding de categoricas

```python
le_sex = LabelEncoder()
le_emb = LabelEncoder()

df['Sex'] = le_sex.fit_transform(df['Sex'])
df['Embarked'] = le_emb.fit_transform(df['Embarked'])

df.head()
```

## Normalizacion y escalamiento

```python
qt = QuantileTransformer(output_distribution='normal', n_quantiles=500)
df[['Age', 'Fare']] = qt.fit_transform(df[['Age', 'Fare']])

scaler = MinMaxScaler()
columnas_feat = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
df[columnas_feat] = scaler.fit_transform(df[columnas_feat])

df.head()
```

## Guardar dataset transformado

```python
df.to_csv('./data/titanic_features.csv', index=False)
df.shape
```
