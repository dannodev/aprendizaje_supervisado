# 2. Analisis exploratorio (EDA)
Objetivo: entender la distribucion de variables y su relacion con la supervivencia.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
```

## Carga de datos

```python
df = pd.read_csv('./data/titanic_clean.csv')
df.head()
```

## Resumen rapido

```python
df.info()

df.describe(include='all').T
```

## Variable objetivo

```python
df['Survived'].value_counts(normalize=True)

plt.figure(figsize=(4, 3))
sns.countplot(data=df, x='Survived')
plt.title('Distribucion de Survived')
plt.show()
```

## Relacion con variables categoricas

```python
plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survived por Sex')
plt.show()

plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survived por Pclass')
plt.show()

plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='Embarked', hue='Survived')
plt.title('Survived por Embarked')
plt.show()
```

## Distribuciones numericas

```python
plt.figure(figsize=(5, 3))
sns.histplot(data=df, x='Age', hue='Survived', kde=True, bins=30)
plt.title('Age vs Survived')
plt.show()

plt.figure(figsize=(5, 3))
sns.histplot(data=df, x='Fare', hue='Survived', kde=True, bins=30)
plt.title('Fare vs Survived')
plt.show()
```
