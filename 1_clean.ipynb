# 1. Limpieza de datos
Objetivo: preparar un dataset limpio con las columnas necesarias para el modelado.

```python
import pandas as pd
import numpy as np
```

## Carga de datos

```python
df = pd.read_csv('./data/titanic.csv')
df.head()
```

## Seleccion de columnas
Nos quedamos con las variables usadas en el modelado.

```python
columnas = [
    'Survived', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked'
 ]
df = df[columnas]
df.head()
```

## Valores faltantes

```python
df.isna().sum()

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df.isna().sum()
```

## Guardar dataset limpio

```python
df.to_csv('./data/titanic_clean.csv', index=False)
df.shape
```
